```python
class Solution:
    def sort012(self, arr):
        n = len(arr)
        l = 0
        r = n - 1
        m = 0
        while m <= r:
            if arr[m] == 0:
                arr[l], arr[m], = arr[m], arr[l]
                l += 1
                m += 1
            elif arr[m] == 1:
                m += 1
            else:
                arr[m], arr[r] = arr[r], arr[m]
                r -= 1
```

```python
class Solution:
    def firstSearch(self, arr, k):
        left = 0
        right = len(arr)-1
        ans = -1
        
        while left<=right:
            middle = left + (right - left) // 2
            
            if arr[middle] == k:
                ans = middle
                right = middle-1
            elif arr[middle] < k:
                left = middle + 1
            else:
                right = middle - 1
        return ans
        
```
