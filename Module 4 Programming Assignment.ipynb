```python
#16.4
import sqlite3
conn = sqlite3.connect('books.db')
curs = conn.cursor()
curs.execute('DROP TABLE IF EXISTS books')
curs.execute('''CREATE TABLE books
             (title TEXT,
             author TEXT,
             year INTEGER)''')
conn.commit()
conn.close()
```

```python
#16.5
import csv, sqlite3

conn = sqlite3.connect('books.db')
curs = conn.cursor()
csv_file = open('books2.csv', 'r', encoding='utf-8')
reader = csv.DictReader(csv_file)
for row in reader:
    curs.execute('INSERT INTO books VALUES (?, ?, ?)',
                   (row['title'], row['author'], int(row['year'])))
conn.commit()
csv_file.close()
conn.close()
```
